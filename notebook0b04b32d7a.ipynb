```python
import pandas as pd
import os
import glob

# Find all CSV files in the Kaggle input folder
csv_files = glob.glob("/kaggle/input/**/*.csv", recursive=True)

print("CSV files found:", csv_files)

# Load the first CSV file
if csv_files:
    file_path = csv_files[0]
    df = pd.read_csv(file_path, encoding="latin1")

    print("\nSUCCESS! Dataset loaded.")
    print("Dataset dimensions:", df.shape)

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nFirst five rows:")
    display(df.head())
else:
    print("No CSV files found. Please check Add Input.")
```

```python
# STEP 1: Calculate Key Business Metrics

# Total Sales
total_sales = df["Sales"].sum()

# Total Profit
total_profit = df["Profit"].sum()

# Total Number of Unique Orders
total_orders = df["Order ID"].nunique()

# Display Results
print("=" * 40)
print("       E-COMMERCE SALES ANALYSIS")
print("=" * 40)

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Unique Orders: {total_orders:,}")

print("=" * 40)
```

```python
# STEP 2: Sales Analysis by Product Category

category_sales = df.groupby("Category")["Sales"].sum()
category_sales = category_sales.sort_values(ascending=False)

print("TOTAL SALES BY CATEGORY")
print("-" * 35)

display(category_sales.to_frame(name="Total Sales"))

# Find the category with the highest sales
best_category = category_sales.idxmax()
highest_sales = category_sales.max()

print("\nCategory with the Highest Sales:", best_category)
print(f"Sales Amount: ${highest_sales:,.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

category_sales.sort_values().plot(
    kind="bar",
    title="Total Sales by Product Category"
)

plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# STEP 3: Sales Analysis by Region

region_sales = df.groupby("Region")["Sales"].sum()
region_sales = region_sales.sort_values(ascending=False)

print("TOTAL SALES BY REGION")
print("-" * 35)

display(region_sales.to_frame(name="Total Sales"))

# Identify the highest-sales region
best_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("\nRegion with the Highest Sales:", best_region)
print(f"Sales Amount: ${highest_region_sales:,.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

region_sales.sort_values().plot(
    kind="bar",
    title="Total Sales by Region"
)

plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Convert Order Date into a proper date format

df["Order Date"] = pd.to_datetime(df["Order Date"])

# Calculate monthly sales

monthly_sales = df.set_index("Order Date")["Sales"].resample("MS").sum()

# Display monthly sales

print("MONTHLY SALES ANALYSIS")
print("-" * 35)
display(monthly_sales.to_frame(name="Total Sales"))

# Create a line chart

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
monthly_sales.plot(
kind="line",
marker="o",
title="Monthly Sales Trend"
)

plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.tight_layout()
plt.show()
```

```python
# Calculate total sales for each product

top_products = df.groupby("Product Name")["Sales"].sum().nlargest(10)

# Display the top 10 products

print("TOP 10 PRODUCTS BY SALES")
print("-" * 40)
display(top_products.to_frame(name="Total Sales"))

# Create a horizontal bar chart

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
top_products.sort_values().plot(
kind="barh",
title="Top 10 Products by Sales"
)

plt.xlabel("Total Sales ($)")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```
